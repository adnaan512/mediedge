# MediEdge — Exploratory Data Analysis

Quick exploration of DermaMNIST and PathMNIST datasets.

```python
import sys
sys.path.insert(0, '..')

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from data.dataloader import get_loaders, get_dataset_info
```

```python
# Load DermaMNIST
info = get_dataset_info('dermamnist')
print(info)

train_loader, val_loader, test_loader = get_loaders(
    dataset_name='dermamnist',
    batch_size=64,
    data_dir='../data/cache'
)
```

```python
# Visualize sample images
images, labels = next(iter(train_loader))
label_names = info['label_names']

fig, axes = plt.subplots(4, 8, figsize=(16, 8))
for i, ax in enumerate(axes.flat):
    if i >= len(images):
        break
    img = images[i].permute(1, 2, 0).numpy()
    img = (img * 0.5 + 0.5).clip(0, 1)  # Denormalize
    ax.imshow(img)
    ax.set_title(label_names[labels[i].item()], fontsize=7)
    ax.axis('off')

plt.suptitle('DermaMNIST Sample Images', fontsize=14)
plt.tight_layout()
plt.savefig('../results/dermamnist_samples.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Class distribution
all_labels = []
for _, lbls in train_loader:
    all_labels.extend(lbls.squeeze().tolist())

unique, counts = np.unique(all_labels, return_counts=True)
class_names = [label_names[i] for i in unique]

plt.figure(figsize=(10, 5))
sns.barplot(x=class_names, y=counts)
plt.title('DermaMNIST — Class Distribution (Train)')
plt.xlabel('Class')
plt.ylabel('Count')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.savefig('../results/dermamnist_class_dist.png', dpi=150, bbox_inches='tight')
plt.show()

print('\nClass distribution:')
for name, count in zip(class_names, counts):
    print(f'  {name:30s}: {count}')
```
